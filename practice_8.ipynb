Mulitindex - wielopoziomowy index. Może być wierszowy lub kolumnowy. Pozwala na bardziej złożone struktury danych.

```python
import pandas as pd

# Twoje dane wyjściowe
data = {
    'Service': ['Netflix', 'Spotify', 'iCloud', 'HBO Max', 'YouTube Premium', 'Disney+'],
    'Monthly Cost (USD)': [29.99, 45.00, 15.00, 60.00, 10.00, 12.99],
    'Category': ['Entertainment', 'Music', 'Cloud Storage', 'Entertainment', 'Video', 'Entertainment']
}

df = pd.DataFrame(data)

# Tworzymy MultiIndex: najpierw Kategoria, potem Usługa
# Ważne: warto posortować indeks, aby Pandas działał szybciej i bez błędów!
df_multi = df.set_index(['Category', 'Service']).sort_index()

print("Tabela z MultiIndexem:")
display(df_multi)
```

```python
#pobranie danych dla poziomu kategorii 'Entertainment' wartości Spotify
print("\nużycie funkcji .xs,  dla kategorii 'Music' do znalezienia Spotify':")
display(df_multi.xs(('Music', 'Spotify'))) # wejdź do poziomu Music i znajdź Spotify
display(df_multi.xs('Spotify', level='Service')) # wejdź do poziomu Service i znajdź Spotifyjak bb
```

Groupby - grupowanie danych po określonej kolumnie lub kolumnach. Pozwala na agregację danych i wykonywanie operacji statystycznych na grupach.


```python
import pandas as pd
#przygotowanie danych
data = {
    'Service': ['Netflix', 'Spotify', 'iCloud', 'HBO Max', 'YouTube Premium', 'Disney+'],
    'Monthly Cost (USD)': [29.99, 45.00, 15.00, 60.00, 10.00, 12.99],
    'Category': ['Entertainment', 'Music', 'Cloud Storage', 'Entertainment', 'Video', 'Entertainment']
}
df = pd.DataFrame(data)
#grupowanie danych po kolumnie 'Category' i sumowanie kosztów
#wybieramy kolumny do grupowania i agregacji Category i Monthly Cost (USD)
cost_by_category = df.groupby('Category')['Monthly Cost (USD)'].sum().reset_index() #reset_index() aby uzyskać ładny DataFrame
print("Suma miesięcznych kosztów subskrypcji według kategorii:")
display(cost_by_category)
```

Funkcja .agg aggregate - pozwala na wykonywanie wielu operacji agregujących na grupach danych jednocześnie.

```python
# sprawdzamy sumę, średnią i liczbę subskrypcji w każdej kategorii
stats = df.groupby('Category')['Monthly Cost (USD)'].agg(['sum', 'mean', 'count']).reset_index()
print("\nStatystyki miesięcznych kosztów subskrypcji według kategorii:")
display(stats)
```

```python
# poprawione wyświetlanie danych
print("\nPoprawione wyświetlanie danych:")
stats = df.groupby('Category')['Monthly Cost (USD)'].agg(
    Suma=('sum'),
    Średnia=('mean'),
    Liczba_Subskrypcji=('count')
    #Kategoria=('Category')
).reset_index()
stats = stats.rename(columns={'Category': 'Kategoria'}) # tu sobie zmieniam nazwe kolumny

# jeszcze nie podoba mi się numerowanie wierszy
#stats.index += 1  # zaczynamy numerowanie od 1 zamiast 0 - to jednak ma wady bo się nie będzie nadpisywać przy kolejnych uruchomieniach tyko liczyć dalej

# To zawsze zresetuje indeks do 1, 2, 3, nawet jeśli uruchomisz to 100 razy
#stats.index = range(1, len(stats) + 1)

# --- KROK 3: Numerowanie wierszy 1., 2., 3... --- To jest super fajne! Bardziej finezyjne.
stats.index = [f"{i}." for i in range(1, len(stats) + 1)]
stats.index.name = 'Lp.'

display(stats)
```

```python
import pandas as pd

# Twoje dane z nową kolumną (jako tekst)
data = {
    'Service': ['Netflix', 'Spotify', 'iCloud', 'HBO Max', 'YouTube Premium', 'Disney+'],
    'Monthly Cost (USD)': [29.99, 45.00, 15.00, 60.00, 10.00, 12.99],
    'Renewal_Date': ['2026-01-15', '2026-02-10', '2026-01-20', '2026-03-05', '2026-02-15', '2026-01-05']
}
df = pd.DataFrame(data)

# KONWERSJA: Zamieniamy tekst na prawdziwy format daty
df['Renewal_Date'] = pd.to_datetime(df['Renewal_Date'])

print("Typ kolumny Renewal_Date po konwersji:", df['Renewal_Date'].dtype)
```

```python
#wyciągnięcie roku miesiąca i dnia z daty odnowienia
df['Rok_Odnowienia'] = df['Renewal_Date'].dt.year
df['Miesiąc_Odnowienia'] = df['Renewal_Date'].dt.month
df['Dzień_Odnowienia'] = df['Renewal_Date'].dt.day

#Tłumaczenie na polski dni tygodnia
days_pl = {
    'Monday': 'Poniedziałek',
    'Tuesday': 'Wtorek',
    'Wednesday': 'Środa',
    'Thursday': 'Czwartek',
    'Friday': 'Piątek',
    'Saturday': 'Sobota',
    'Sunday': 'Niedziela'
}
df['Dzień_Tygodnia_Odnowienia'] = df['Renewal_Date'].dt.day_name().map(days_pl) # Mapowanie na polskie nazwy dni tygodnia
display(df[['Service', 'Renewal_Date', 'Dzień_Tygodnia_Odnowienia']])
```

```python
import pandas as pd

# 1. Tworzymy bazowe dane DataFrame
data = {
    'Service': ['Netflix', 'Spotify', 'iCloud', 'HBO Max', 'YouTube Premium', 'Disney+'],
    'Monthly Cost (USD)': [29.99, 45.00, 15.00, 60.00, 10.00, 12.99],
    'Category': ['Entertainment', 'Music', 'Cloud Storage', 'Entertainment', 'Video', 'Entertainment'],
    'Renewal_Date': ['2026-01-15', '2026-02-10', '2026-01-20', '2026-03-05', '2026-02-15', '2026-01-05']
}
df = pd.DataFrame(data)

# 2. Dodajemy brakujące kolumny z datami
df['Renewal_Date'] = pd.to_datetime(df['Renewal_Date'])
df['Miesiąc_Odnowienia'] = df['Renewal_Date'].dt.month

# 3. Teraz wykonujemy grupowanie po więcej niż jednej kolumnie
# Grupujemy po 'Category' i 'Miesiąc_Odnowienia', obliczamy sumę i liczbę usług
raport_multi = df.groupby(['Category', 'Miesiąc_Odnowienia'])['Monthly Cost (USD)'].agg(
    Suma_Wydatków='sum',
    Liczba_Usług='count'
).reset_index()

# 4. Poprawiamy nazwy na polskie
raport_multi = raport_multi.rename(columns={
    'Category': 'Kategoria',
    'Miesiąc_Odnowienia': 'Miesiąc'
})

# 5. Twoje numerowanie 1., 2., 3...
raport_multi.index = [f"{i}." for i in range(1, len(raport_multi) + 1)]
raport_multi.index.name = 'Lp.'

print("Naprawiony raport wielopoziomowy:")
display(raport_multi)
```

MERGE, JOIN, CONCATENATE czyli łączenie danych z różnych źródeł.

```python
import pandas as pd

# Dane ze stycznia
df_jan = pd.DataFrame({
    'Service': ['Netflix', 'Spotify'],
    'Cost': [29.99, 45.00]
})

# Dane z lutego
df_feb = pd.DataFrame({
    'Service': ['iCloud', 'Spotify'],
    'Cost': [15.00, 45.00]
})

# Łączymy je w jedną długą tabelę
df_all = pd.concat([df_jan, df_feb], ignore_index=True) #ignore_index=True resetuje indeksy w nowej tabeli

#wyświetlamy najpierw styczniowe i lutowe
print("Dane ze stycznia:")
display(df_jan)

print("Dane z lutego:")
display(df_feb)

print("Połączone wiersze (concat):")
display(df_all)
```

Łacznie relacyjne - łączenie danych na podstawie wspólnych kluczy (kolumn) w różnych tabelach.
MERGE - łączenie danych na podstawie wspólnych kolumn (kluczy) w różnych tabelach.

```python
# Tabela 1: Ceny usług
df_prices = pd.DataFrame({
    'Service': ['Netflix', 'Spotify', 'iCloud'],
    'Cost': [29.99, 45.00, 15.00]
})

# Tabela 2: Kategorie usług
df_cats = pd.DataFrame({
    'Service': ['Netflix', 'Spotify', 'iCloud', 'Disney+'],
    'Category': ['Ent.', 'Music', 'Cloud', 'Ent.']
})

# Łączymy po kolumnie 'Service'
# how='inner' - zostawia tylko to, co jest w OBU tabelach
df_merged = pd.merge(df_prices, df_cats, on='Service', how='outer') # tu mamy parametr o którym mowa poniżej

print("Połączone informacje (merge):")
display(df_merged)

#inner - łączenie wewnętrzne - tylko wspólne wiersze z obu tabel
#left - wszystkie z lewej, + to co pasuje z prawej
#outer - wszystkie z obu tabel, brakujące wypełnione NaN (pojawi sie Disney+ z brakiem ceny)
```

JOIN - łączenie danych na podstawie indeksów w różnych tabelach.

```python
# Ustawiamy 'Service' jako indeks w obu tabelach
df_prices_idx = df_prices.set_index('Service')
df_cats_idx = df_cats.set_index('Service')

# Łączymy po indeksie
df_joined = df_prices_idx.join(df_cats_idx)

#najpierw wyświetlamy obie tabele
print("Tabela cen (z indeksem):")
display(df_prices_idx)
print("Tabela kategorii (z indeksem):")
display(df_cats_idx)

# axis=1 oznacza: sklejaj kolumny obok siebie (poziomo)
print("Tabele sklejone technicznie (poziomo):")
display(pd.concat([df_prices_idx, df_cats_idx], axis=1))

print("Połączone po indeksie (join):")
display(df_joined)
```

```python
import pandas as pd

# 1. Przygotowujemy nasze dane
data = {
    'Usługa': ['Netflix', 'Spotify', 'iCloud', 'Disney+'],
    'Koszt': [29.99, 45.00, 15.00, 37.99]
}
df = pd.DataFrame(data)

# 2. Tworzymy plik z silnikiem xlsxwriter
with pd.ExcelWriter('Raport_z_Walidacja.xlsx', engine='xlsxwriter') as writer:
    df.to_excel(writer, sheet_name='Kalkulator', index=False) #index=False aby nie zapisywać indeksu
    
    workbook  = writer.book
    worksheet = writer.sheets['Kalkulator']

    # --- INSTRUKCJA Z KURSU: WPISYWANIE TEKSTU W A16 ---
    # Dodajemy prośbę do użytkownika
    label = "Podaj liczbę osób dzielących koszty (całkowita > 0):"
    res = worksheet.write_string('A16', label)
    
    # Wyjaśnienie kodów błędu z kursu:
    if res == 0:
        print("Tekst w A16 wpisany poprawnie!")
    elif res == -1:
        print("Błąd: Komórka poza zakresem arkusza!")

    # --- INSTRUKCJA Z KURSU: WALIDACJA W B16 ---
    # Zezwalamy tylko na liczby całkowite (integer) większe niż 0
    worksheet.data_validation('B16', {
        'validate': 'integer',
        'criteria': '>',
        'value': 0,
        'input_title': 'Wprowadź dane:',
        'input_message': 'Wpisz liczbę osób (np. 1, 2, 3...)',
        'error_message': 'Błąd! Musisz podać liczbę całkowitą większą od zera.'
    })

    # --- INSTRUKCJA Z KURSU: ROZSZERZANIE KOLUMNY A ---
    # Ustawiamy szerokość kolumny A na 45 jednostek, aby tekst był widoczny
    worksheet.set_column('A:A', 45)

    # DODATKOWO (dla efektu): Walidacja z listy (o której wspomina kurs)
    # W komórce B17 pozwolimy wybrać kategorię tylko z listy
    worksheet.write_string('A17', "Wybierz priorytet płatności:")
    worksheet.data_validation('B17', {
        'validate': 'list',
        'source': ['Niski', 'Średni', 'Wysoki', 'Krytyczny']
    })

import os
os.startfile('Raport_z_Walidacja.xlsx')
```

```python
import pandas as pd
import requests

url = 'https://pl.wikipedia.org/wiki/Miasta_w_Polsce_(statystyki)'

# 1. Definiujemy "tożsamość" przeglądarki (User-Agent)
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
} # ważne, aby strona nas nie zablokowała jako bota podszywamy sie pod prawdziwą przeglądarkę

# 2. Pobieramy zawartość strony za pomocą requests
response = requests.get(url, headers=headers)

# 3. Przekazujemy tekst strony (HTML) do Pandas
# Zamiast adresu URL, podajemy response.text
tabele = pd.read_html(response.text)

print(f"Sukces! Znalazłem {len(tabele)} tabel(ę).")

# Wybieramy pierwszą tabelę
miasta_df = tabele[0]
display(miasta_df.head())
```

Dokładam formatowanie

```python
#Formatowanie kolumny Lp.
miasta_df.index = [f"{i}." for i in range(1, len(miasta_df) + 1)]
miasta_df.index.name = 'Lp.'

display(miasta_df.head())
```

Teraz dokładamy formatowanie do naszego excela.

```python
with pd.ExcelWriter('Miasta_w_Polsce.xlsx', engine='xlsxwriter') as writer:
    # Zapisujemy główny DataFrame
    miasta_df.to_excel(writer, sheet_name='Ranking Miast', index=True) #index=True aby zapisywać indeks
    
    workbook  = writer.book
    worksheet = writer.sheets['Ranking Miast']

    #Tworzenie formatu dla nagłówków pogrubienie i środkowanie
    header_format = workbook.add_format({
        'bold': True,
        'text_wrap': True,
        'valign': 'center',
        'align': 'center',
        'border': 1
    })
    # Zastosowanie formatu do nagłówków
    for col_num, value in enumerate(miasta_df.columns.values):
        worksheet.write(0, col_num + 1, value, header_format)  # +1 bo indeksy w Excelu zaczynają się od 1 dla kolumny B

print("Plik 'Miasta_w_Polsce.xlsx' został utworzony z formatowaniem nagłówków.")
os.startfile('Miasta_w_Polsce.xlsx')
```
